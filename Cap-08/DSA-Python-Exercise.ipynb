```python
from math import sqrt

class Rocket():
     
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y
        
    def move_rocket(self, x_increment=0, y_increment=1):
        self.x += x_increment
        self.y += y_increment
        
    def print_rocket(self):
        print(self.x, self.y)
```

```python
rocket = Rocket(1,2)

print(hasattr(rocket, "x"))
print(getattr(rocket, "x"))
rocket.print_rocket()
```

```python
class Person:
    def __init__(self, name: str, city: str, phone: str, email: str):
        self.__name = name
        self.__city = city
        self.__phone = phone
        self.__email = email

    @property
    def name(self) -> str:
        return self.__name

    @name.setter
    def name(self, name: str):
        self.__name = name

    @property
    def city(self) -> str:
        return self.__city

    @city.setter
    def city(self, city: str):
        self.__city = city

    @property
    def phone(self) -> str:
        return self.__phone

    @phone.setter
    def phone(self, phone: str):
        self.__phone = phone

    @property
    def email(self) -> str:
        return self.__email

    @email.setter
    def email(self, email: str):
        self.__email = email

    def __str__(self) -> str:
        return f"Name: {self.__name}, City: {self.__city}, Phone: {self.__phone}, E-mail: {self.__email}"

class ServicePerson:

    def __init__(self, person: Person, salary: float):
        self.__person = person
        self.__salary = salary

    @property
    def salary(self) -> float:
        return self.__salary

    @salary.setter
    def salary(self, salary):
        self.__salary = salary

    def increase_salary(self, amount: float):
        self.__salary += amount

    def __str__(self):
        return f"Object: {self.__person};\nSalary: {self.__salary}"
```

```python
person1 = Person("caio", "recife", "(81)997447053", "caio@gmail.com")

print(person1)

person1.name = "lucas"

print(f"String format: {person1};\n Get Name atribute: {person1.name};\n Get phone: {person1.phone}")

servicePerson = ServicePerson(person1, 4200.00)

print(servicePerson)

servicePerson.increase_salary(1000)

print(servicePerson)
```

```python
class Smartphone:

    def __init__(self, size: str, interface: str, battery: str):
        self.__size = size + "px"
        self.__interface = interface
        self.__battery = battery + "mAh"

    @property
    def size(self) -> str:
        return self.__size

    @size.setter
    def size(self, size: str):
        self.__size = size + "px"

    @property
    def interface(self) -> str:
        return self.__interface

    @interface.setter
    def interface(self, interface: str):
        self.__interface = interface

    @property
    def battery(self) -> str:
        return self.__battery

    @battery.setter
    def battery(self, bateria: str):
        self.__battery = battery + "mAh"

    
    def battery_verification(self) -> str:
        return f"Battery on {self.__battery} is charged."

    def __str__(self):
        return f"Size: {self.__size}\nInterface: {self.__interface}\nBattery: {self.__battery}"


class Mp3Player(Smartphone):

    def __init__(self, size: str, interface: str, battery: str, capacity: str):
        super().__init__(size, interface, battery)
        self.__capacity = capacity

    @property
    def capacidade(self) -> str:
        return self.__capacidade

    @capacidade.setter
    def capacidade(self, capacidade: str):
        self.__capacidade = capacidade

    def __str__(self):
        return f"{super().__str__()}\nCapacidade: {self.__capacity}"

```

```python
smartphone = Smartphone("1080", "Touchscreen", "4000")
print("Smartphone:")
print(smartphone)
print()

mp3_player = Mp3Player("720", "LCD", "4000", "64GB")
print("MP3 Player:")
print(mp3_player)
print(mp3_player.battery_verification())
```
