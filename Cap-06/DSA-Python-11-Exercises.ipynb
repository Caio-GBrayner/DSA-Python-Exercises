```python
lista = [0, 1, 2, 3, 4]

results = list(map(lambda x: (x ** 2, x ** 3),lista))

print(results,"\n")

cubo =[row[0] for row in results]
quadrado =[row[1] for row in results]

lista_completa = [[row[0] for row in results], [row[1] for row in results]]

print(f"Cubo: {cubo}\nQuadrado: {quadrado}\nLista Completa: {lista_completa}")
```

```python
listaA = [1, 2, 3, 5, 7, 9]
listaB = [2, 3, 5, 6, 7, 8]

listaC = listaA + listaB

print(listaC)  
print(type(listaC)) 

def list_comparison(lista) -> list:
    repetidos = [] 
    for i, x in enumerate(lista):
        if x in lista[i + 1:]:
            if x not in repetidos:
                repetidos.append(x)
    return repetidos

print(list_comparison(listaC))
```
